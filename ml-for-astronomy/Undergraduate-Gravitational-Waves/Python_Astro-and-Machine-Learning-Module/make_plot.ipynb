```python
import numpy as np
import matplotlib.pyplot as plt
from pycbc.waveform import get_td_waveform
```

```python
def extract_merger_window(h, window_size):
    peak = np.argmax(np.abs(h))
    half = window_size // 2

    start = max(0, peak - half)
    end = start + window_size

    if end > len(h):
        end = len(h)
        start = end - window_size

    h_win = h[start:end]

    if len(h_win) < window_size:
        h_win = np.pad(h_win, (0, window_size - len(h_win)))

    return h_win


fs = 4096
delta_t = 1.0 / fs
f_lower = 30

n_bns = 5
window_size = 4096   # 1 second window

time = np.arange(window_size) * delta_t

# -----------------------------
# BNS waveforms
# -----------------------------
fig = plt.figure()

for _ in range(n_bns):
    m1 = np.random.uniform(1.2, 1.6)
    m2 = np.random.uniform(1.2, 1.6)

    hp, _ = get_td_waveform(
        approximant="IMRPhenomD_NRTidal",
        mass1=m1,
        mass2=m2,
        delta_t=delta_t,
        f_lower=f_lower
    )

    h = extract_merger_window(hp.numpy(), window_size)


    numerator = (m1 * m2)**(3.0/5.0)
    denominator = (m1 + m2)**(1.0/5.0)
    chirp_mass = round(numerator / denominator, 2)

    plt.plot(time, h, lw=1.5, label=f"M_chirp = {chirp_mass} M_sol")

plt.ylabel("Strain")
plt.legend(loc='upper left', fontsize=9)
plt.xlabel("Time (s)")
plt.xlim(0.2,0.52)
plt.tight_layout()
plt.show()
```
